```python
"""

fields.py — electric and magnetic field definitions.

Units are normalized: q = m = B0 = 1, so the cyclotron frequency is 1
and time is measured in units of 1/omega_c.

Every field function takes a position x (shape (3,)) and time t,
and returns a vector of shape (3,).
"""
import numpy as np                              # vectors, cross products, arrays of particles


def uniform_B(B0=1.0):
    """Uniform magnetic field along z."""
    def B(x, t):
        return np.array([0.0, 0.0, B0])
    return B

def zero_E():
    """No electric field."""
    def E(x, t):
       return np.zeros(3)
    return E

def uniform_E(E0=1.0):
    """Uniform magnetic field along z."""
    def E(x, t):
        return np.array([E0, 0.0, 0.0])
    return E

    


def mirror_B(B0=1.0, L=10.0):
    """Magnetic mirror along z: weak in the middle, strong at the ends.

    B_z = B0 (1 + z^2/L^2)          -> B0 at z=0, 2*B0 at z=±L
    B_x = -B0 x z / L^2             (these two small radial parts
    B_y = -B0 y z / L^2              make div B = 0 and squeeze the field lines)
    """
    def B(x, t):
        X, Y, Z = x
        return B0 * np.array([-X * Z / L**2,
                              -Y * Z / L**2,
                              1 + Z**2 / L**2])
    return B
```
